# 🚀 NoxMedia — Telegram Session String Generator
### 100% Safe, Open-Source & Runs in your own Google Cloud Sandbox

This notebook allows you to generate your **Telegram Session String** for [NoxMedia](https://github.com/TheAceOfficials/NoxMedia) in 1 minute without needing any PC software or third-party bots.

**Instructions:**
1. Tap the **▶ Run** button on the code cell below.
2. Enter your `API ID`, `API HASH`, and `Phone Number` when prompted.
3. Enter the official Telegram OTP code.
4. Your **Session String** will be generated and automatically sent to your Telegram **Saved Messages** chat!

```python
# Step 1: Install Pyrogram & TgCrypto
!pip install -q pyrogram tgcrypto

import asyncio
from pyrogram import Client

async def generate_session():
    print("=" * 50)
    print("  NoxMedia Session String Generator")
    print("=" * 50)
    
    api_id_val = input("Enter API ID (from my.telegram.org): ").strip()
    api_hash_val = input("Enter API HASH (from my.telegram.org): ").strip()
    
    if not api_id_val or not api_hash_val:
        print("❌ Error: API ID and API HASH cannot be empty.")
        return
        
    api_id = int(api_id_val)
    api_hash = api_hash_val
    
    print("\nConnecting to official Telegram servers...")
    async with Client(":memory:", api_id=api_id, api_hash=api_hash, in_memory=True) as app:
        session_str = await app.export_session_string()
        
        print("\n" + "=" * 50)
        print("  ✓ SUCCESS! YOUR SESSION STRING TOKEN:")
        print("=" * 50)
        print(f"\n{session_str}\n")
        print("=" * 50)
        
        try:
            await app.send_message("me", f"**Your NoxMedia Session String:**\n\n`{session_str}`")
            print("✓ Also sent directly to your Telegram 'Saved Messages' chat!")
        except Exception:
            pass
            
        print("\nCopy the string above and paste it into NoxMedia app!")

# Run generator
try:
    loop = asyncio.get_event_loop()
    loop.run_until_complete(generate_session())
except RuntimeError:
    asyncio.run(generate_session())
```
